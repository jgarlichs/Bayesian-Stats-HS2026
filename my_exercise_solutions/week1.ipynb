```python
# 2)
def analytic_pi(k):
    sum = 0
    for i in range(k):
        sum += (-1)**i/(2*i +1)
    return 4*sum
```

```python
def mc_pi(k):
    counts = 0
    points = np.random.uniform(0, 1, size=np.array([k,k]))
    for j in range(k):
        x = points[j,0]
        y = points[j,1]
        if np.sqrt(x**2 + y**2) < 1:
            counts += 1
    return 4*counts/k
```

```python
# 4)
ns = np.linspace(2,100, 101)

series = np.array([analytic_pi(int(i)) for i in ns])
mc = np.array([mc_pi(int(i)) for i in ns])
```

```python
plt.figure(dpi=150)
plt.plot(ns, series, label="Analytic series")
plt.plot(ns, mc, label="Monte Carlo")
plt.hlines(np.pi, ns.min(), ns.max(), label="Numpy pi", color="grey", linestyle="--")
plt.legend()
plt.show()
```
